### Practical 1: Count Lines, Words and Characters

**Problem Definition:**  
Write a Python program to read text from a file and count the total number of lines, words, and characters. The result should be written to an output file.


```python
line_count=0
word_count=0
char_count=0
with open('new.txt','r+') as fp:
    for l in fp:
        line_count += 1
        word_count += len(l.split())
        char_count += len(l)
with open('output.txt','w') as fp1:
    fp1.write(f'no of lines:{str(line_count)}\n')
    fp1.write(f'no of words:{str(word_count)}\n')
    fp1.write(str(char_count))
```


### Practical 2: Search a Word in a File

**Problem Definition:**  
Write a Python program to search a given word in a text file and count its occurrences. Store the result in an output file.


```python
def frequency(filename,word):
    count = 0
    with open(filename,'r') as fp:
        for line in fp:
            words = line.split()
            for w in words:
                if w.lower() == word.lower():
                    count += 1
    return count
file_name = 'new.txt'
search_word = input("enter word to search:")
result = frequency(file_name,search_word)
with open('output.txt','w') as fp1:
    fp1.write(f"Occurrence of '{search_word}' ={result}")
```


### Practical 3: Student Marks and Average

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Store marks in a list, calculate the average using a user-defined function, and write student details along with average marks to an output file.


```python
def calc_avg(marks):
    total = 0
    for m in marks:
        total += m
    return total / len(marks)
    
fin = open("marks.txt","r")
fout = open("res.txt","w")
data=fin.readlines()
for line in data:
    data = line.split()
    # print(data)
    roll = data[0]
    name = data[1]

    marks = []
    for i in range(2,len(data)):
        marks.append(int(data[i]))
        
    avg = calc_avg(marks)
    
    fout.write("Roll No: " + roll + "\n")
    fout.write("Name: " + name + "\n")
    fout.write("Marks: " + str(marks) + "\n")
    fout.write("Average Marks: " + str(avg) + "\n\n")

fin.close()
fout.close()
```


### Practical 5: Remove Duplicate Words using Set

**Problem Definition:**  
Write a Python program to read words from a file and remove duplicate words using a set. Store the unique words in an output file.


```python
with open('file1.txt','r') as fp:
    content = fp.read()
words = content.split()
unique = set(words)

with open("output1.txt", "w") as outfile:
    for word in unique:
        outfile.write(word + " ")

print("Duplicate words removed successfully. Check output.txt file.")
```


### Practical 6: Word Frequency using Dictionary

**Problem Definition:**  
Write a Python program to count the frequency of each word in a text file using a dictionary and write the result to an output file.


```python
with open("file1.txt", "r") as infile:
    text = infile.read()

text = text.lower()

words = text.split()

word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

with open("output.txt", "w") as outfile:
    for word, count in word_count.items():
        outfile.write(f"{word} : {count}\n")

print("Word frequency counted successfully. Check output.txt file.")
```


### Practical 7: Student Result and Grade Calculation

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Use a user-defined function to determine grade (A/B/C/Fail) and write the student result to an output file.

**Grade Criteria:**  
- Marks ≥ 75 : Grade A  
- Marks ≥ 60 : Grade B  
- Marks ≥ 40 : Grade C  
- Marks < 40 : Fail


```python
def get_grade(marks):
    if marks >= 75:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "Fail"

with open("Students.txt", "r") as infile, open("result.txt", "w") as outfile:
    
    for line in infile:
        data = line.strip().split()
        
        roll_no = data[0]
        name = data[1]
        marks = int(data[2])
        
        grade = get_grade(marks)
        
        outfile.write(f"Roll No: {roll_no}, Name: {name}, Marks: {marks}, Grade: {grade}\n")

print("Results processed successfully. Check result.txt file.")
```
