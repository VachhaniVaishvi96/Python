## Practical 1: Student Marks Analyzer

### Problem Definition
Store student data as a list of tuples. Remove duplicate marks, calculate average marks and store result in a dictionary.

### Sample Input
```python
students = [
    (1, "Amit", [78, 85, 90, 85]),
    (2, "Neha", [88, 92, 79, 88]),
    (3, "Ravi", [70, 75, 80, 75])
]
```

### Sample Output
```
{1: ('Amit', 84.33), 2: ('Neha', 86.33), 3: ('Ravi', 75.0)}

```

````python
students = [(1, "Amit", [78, 85, 90, 85]), (2, "Neha", [88, 92, 79, 88]), (3, "Ravi", [70, 75, 80, 75])]
res = {}
for a,b,c in students:
    unique=set(c)
    avg_marks= round(sum(unique)/len(unique),2)
    res[a]=(b,avg_marks)
print(res)
````

## Practical 2: Inventory Management

### Problem Definition
Remove out-of-stock items and calculate total inventory value.

### Sample Input
```python
inventory = {
    'pen': (10, 50),
    'book': (100, 0),
    'pencil': (5, 100)
}
```

### Sample Output
```
{'pen': (10, 50), 'pencil': (5, 100)}
Total Value: 1000
```

````python
inventory = {
    'pen': (10, 50),
    'book': (100, 0),
    'pencil': (5, 100)
}
updated_inventory = {item: (price, qty) 
                     for item, (price, qty) in inventory.items() 
                     if qty > 0}

total_value = sum(price * qty for price, qty in updated_inventory.values())

print(updated_inventory)
print("Total Value:", total_value)
````

## Practical 3: Email Domain Extractor

### Problem Definition
Extract email domains and count occurrences.

### Sample Input
```python
emails = ['a@gmail.com', 'b@yahoo.com', 'c@gmail.com']
```

### Sample Output
```
{'gmail.com': 2, 'yahoo.com': 1}
```

````python
emails = ['a@gmail.com', 'b@yahoo.com', 'c@gmail.com']

domain_count = {}

for email in emails:
    domain = email.split("@")[1]
    domain_count[domain] = domain_count.get(domain, 0) + 1

print(domain_count)
````

## Practical 4: Shopping Cart Billing

### Problem Definition
Calculate item-wise cost and total bill.

### Sample Input
```
cart = [('apple',2), ('banana',3)]
prices = {'apple':50, 'banana':10}
```

### Sample Output
```
{'apple': 100, 'banana': 30}
Total Bill: 130
```

````python
cart = [('apple', 2), ('banana', 3)]
prices = {'apple': 50, 'banana': 10}

bill = {}
total = 0

for item, qty in cart:
    cost = prices[item] * qty
    bill[item] = cost
    total += cost

print(bill)
print("Total Bill:", total)
````

## Practical 5: Character Position Index

### Problem Definition
Store positions of each character in a string.

### Sample Input
```python
s = "banana"
```

### Sample Output
```
{'b': [0], 'a': [1, 3, 5], 'n': [2, 4]}
```

````python
s = "banana"

char_positions = {}

for index, ch in enumerate(s):
    if ch in char_positions:
        char_positions[ch].append(index)
    else:
        char_positions[ch] = [index]

print(char_positions)
````

## Practical 6: Survey Data Aggregation

### Problem Definition
Count responses city-wise and choice-wise.

### Sample Input
```python
responses = [
    ('Amit','Rajkot','Yes'),
    ('Neha','Rajkot','No'),
    ('Ravi','Ahmedabad','Yes')
]
```

### Sample Output
```
{'Rajkot': {'Yes': 1, 'No': 1}, 'Ahmedabad': {'Yes': 1}}
```

````python
responses = [
    ('Amit','Rajkot','Yes'),
    ('Neha','Rajkot','No'),
    ('Ravi','Ahmedabad','Yes')
]

result = {}

for name, city, choice in responses:
    if city not in result:
        result[city] = {}
    
    if choice not in result[city]:
        result[city][choice] = 0
    
    result[city][choice] += 1

print(result)
````
