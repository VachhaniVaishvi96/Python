### 08) WAP to create an UDF named frequency to count occurances of the specific word in a given text file.

```python
def frequency(filename, word):
    count = 0
    with open(filename, 'r') as file:
        for line in file:
            words = line.split()
            for w in words:
                if w.lower() == word.lower():
                    count += 1
    return count

file_name = "new.txt"  
search_word = "python" 

result = frequency(file_name, search_word)
print(f"Occurrence of '{search_word}' =", result)
```

### 09) WAP to get the score of five subjects from the user, store them in a file. Fetch those marks and find the highest score.

```python
with open("marks.txt", "w") as file:
    for i in range(1, 6):
        mark = int(input(f"Enter marks of subject {i}: "))
        file.write(str(mark) + "\n")
marks = []

with open("marks.txt", "r") as file:
    for line in file:
        marks.append(int(line.strip()))

highest_score = max(marks)

print("Marks:", marks)
print("Highest score:", highest_score)
```

### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

with open("primenumbers.txt", "w") as file:
    count = 0
    number = 2

    while count < 100:
        if is_prime(number):
            file.write(str(number) + "\n")
            count += 1
        number += 1

print("First 100 prime numbers written to primenumbers.txt")
```
