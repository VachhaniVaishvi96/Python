### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def isPrime(a):
    # num = 2
    count = 0
    for i in range (2,a+1):
        if a % i == 0:
            count +=1
    if (count == 1):
        return 1
    else:
        return 0
isPrime(9)
```

### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
def isPalindrom(s):
    if s == s[::-1]:
        return True
    else:
        return False
isPalindrom("aba")
```

### 07) Write a function to calculate the sum of the first element of each tuples inside the list.

```python
def sum_first_elements(tuple_list):
    total = 0
    for t in tuple_list:
        total += t[0]
    return total
tuple_list = [(1,2),(3,4),(5,6)]
sum_first_elements(tuple_list)
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def Fibonacci(n):
    if n == 0:
        return 1
    elif n == 1:
        return 1
    else:
        return Fibonacci(n-1)+Fibonacci(n-2)
Fibonacci(8)
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
Before = {'a': 10, 'b':20, 'c':30, 'd':40}
After = {}
def invert(Before):
    for k,v in Before.items():
        After[v]=k
    print(After)
invert(Before)
```

### 12) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
s1 = "AbcDEfgh"
def checkCase(s1):
    no_upper = 0
    no_lower = 0
    for ch in s1:
        if ch.isupper():
            no_upper += 1
        elif ch.islower():
            no_lower += 1
    
    print("upper:",no_upper)
    print("lower:",no_lower)
checkCase(s1)
    
```

### 14) For the given list of names of students, extract the names having more that 7 characters. Use filter().

```python
students = ['Ajay', 'Rahul', 'Vaishnavi', 'Pooja', 'Karthikeyan', 'Snehal']
result = list(filter(lambda name: len(name) > 7, students))
print(result)
```

### 15) For the given list of names of students, convert the first letter of all the names into uppercase. use map().

```python
students = ['ajay', 'rahul', 'vaishnavi', 'pooja', 'snehal']
result = list(map(lambda name: name.title(), students))
print(result)
```
